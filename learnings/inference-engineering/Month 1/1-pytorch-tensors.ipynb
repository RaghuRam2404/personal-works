```python
import numpy as np
import torch
```

```python
d = np.array([1,2,3])
t = torch.from_numpy(d)
print(f"{d}, {t}")
tensor_memory = t.data_ptr()
array_memory = d.__array_interface__["data"][0]
assert tensor_memory == array_memory #shared underlying memory
```

```python
t = torch.tensor([1,2,3])
t.type, t.shape, t.dtype, t.device
```

```python
t = torch.tensor([1,2,3], dtype=torch.float64)
t.type, t.shape, t.dtype
```

```python
r1 = torch.rand(size=(1,3))
print(r1)

r2 = torch.rand_like(input = r1)
print(r2, r2.dtype)

o1 = torch.ones(size=(1,3))
print(o1, o1.dtype)

o2 = torch.ones_like(r1)
print(o2)

z1 = torch.zeros(size=(1,3))
print(z1, z1.dtype)

ri1 = torch.randint(low=0, high=10, size=(1,3))
print(ri1)
```

```python
t1 = torch.ones_like(o1)
print(t1.device)

acc_available = torch.accelerator.is_available()
if acc_available:
    acc = torch.accelerator.current_accelerator()
    t2 = t1.to(acc)
    print(t2.device)
    
    #check the memory
    t1_mem = t1.data_ptr()
    t2_mem = t2.data_ptr()
    
    print(t1_mem,' ',t2_mem)
```

```python
t1 = torch.rand(size=(2,3,4))
print(t1)

print(t1[0], t1.shape)
print(t1[:,1])
print(t1[...,-1]) #same as t1[:,:,-1]
```

```python
t = torch.randint(low=0, high=4, size=(2,3))
print(t)
out = torch.cat([t,t,t], dim=1)
print(out, out.shape)
```

```python
t1 = torch.randint(low=0, high=10, size=(3,4))
print(t1 @ t1.T)
print(t1.matmul(t1.T))
```

```python
t1 = torch.randint(low=0, high=10, size=(3,4))
print(t1, t1.sum())
print(t1.sum(dim=0), t1.sum(dim=0, keepdim=True))
print(t1.sum(dim=1), t1.sum(dim=1, keepdim=True))
```

```python
t1 = torch.randint(low=0, high=10, size=(3,4,2))
print(t1, t1.sum())
print(t1.sum(dim=0, keepdim=True).shape, t1.sum(dim=0, keepdim=True).squeeze().shape)
print(t1.sum(dim=1, keepdim=True).shape, t1.sum(dim=1, keepdim=True).squeeze().shape)
print(t1.sum(dim=2, keepdim=True).shape, t1.sum(dim=2, keepdim=True).squeeze().shape)
```
