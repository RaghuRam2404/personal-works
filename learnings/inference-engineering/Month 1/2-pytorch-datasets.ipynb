```python
class CustomDataSet(Dataset):
    def __init__(self, data, target):
        self.data = data
        self.target = target
    
    def __len__(self):
        return len(self.data) if self.data is not None else 0

    def __getitem__(self, idx):
        print(f"Getting the item at idx: {idx}")
        return self.data[idx], self.target[idx]
    
    
xs = torch.randint(0,100, size=(10, 4, 5))
ys = torch.randint(0,2, size=(10,1))

newdset = CustomDataSet(xs, ys)
dl = DataLoader(newdset, batch_size=8, shuffle=False)
for batch_idx, (batch_x, batch_y) in enumerate(dl):
    print(f"\n--- Batch {batch_idx} ---")
    print("Feature batch shape:", batch_x.shape)
    print("Label batch shape:  ", batch_y.shape)
    print("First feature in batch shape:", batch_x[0].shape)
    print("First label in batch:", batch_y[0])
    print("=====================================")
```

```python

data_iterator = iter(dl)


while True:
    batch = next(data_iterator, None)
    
    if batch is None:
        break
    
    batch_x, batch_y = batch

    print("\n--- New batch ---")
    print("Feature batch shape:", batch_x.shape)
    print("Label batch shape:  ", batch_y.shape)

```
