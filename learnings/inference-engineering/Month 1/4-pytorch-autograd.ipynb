```python
import torch
```

```python
x = torch.ones(size=(1,5))
y = torch.zeros(size=(1,3))
w = torch.randn(size=(5,3), requires_grad=True)
b = torch.randn(size=(1,3), requires_grad=True)
z = torch.matmul(x, w) + b
print(f"z.shape={z.shape}")
loss = torch.nn.functional.binary_cross_entropy_with_logits(z, y)
print(loss)
```

```python
w.grad, b.grad
```

```python
loss.backward()
(w.grad, b.grad)
```

```python
x = torch.ones(size=(1,5))
y = torch.zeros(size=(1,3))
w = torch.randn(size=(5,3), requires_grad=True)
b = torch.randn(size=(1,3), requires_grad=True)
z = torch.matmul(x, w) + b
loss = torch.nn.functional.binary_cross_entropy_with_logits(z, y)

print(f"{b.requires_grad=}, {w.requires_grad=}, {z.requires_grad=}")
with torch.no_grad():
    z = torch.matmul(x, w) + b
    print(f"{b.requires_grad=}, {w.requires_grad=}, {z.requires_grad=}")
```

```python
x = torch.ones(size=(1,28*28))
y = torch.ones(size=(1,10))

w1 = torch.randn(size=(28*28, 512), requires_grad=True)
b1 = torch.randn(size=(1, 512), requires_grad=True)
l1 = torch.matmul(x, w1) + b1
h1 = torch.nn.ReLU()(l1)

w2 = torch.randn(size=(512, 256), requires_grad=True)
b2 = torch.randn(size=(1, 256), requires_grad=True)
l2 = torch.matmul(h1, w2) + b2
h2 = torch.nn.ReLU()(l2)

w3 = torch.randn(size=(256, 128), requires_grad=True)
b3 = torch.randn(size=(1, 128), requires_grad=True)
l3 = torch.matmul(h2, w3) + b3
h3 = torch.nn.ReLU()(l3)

w4 = torch.randn(size=(128, 10), requires_grad=True)
b4 = torch.randn(size=(1, 10), requires_grad=True)
l4 = torch.matmul(h3, w4) + b4
h4 = torch.nn.ReLU()(l4)

loss = torch.nn.functional.binary_cross_entropy_with_logits(y, h4)

# Clear the previous gradients (it's not going to be there for the first run)
w1.grad = None
b1.grad = None
w2.grad = None
b2.grad = None
w3.grad = None
b3.grad = None
w4.grad = None
b4.grad = None

loss.backward()
print(w2.grad)
```

```python
# Clear the previous gradients
w1.grad = None
b1.grad = None
w2.grad = None
b2.grad = None
w3.grad = None
b3.grad = None
w4.grad = None
b4.grad = None

h2_new = h2.detach()

w3 = torch.randn(size=(256, 128), requires_grad=True)
b3 = torch.randn(size=(1, 128), requires_grad=True)
l3 = torch.matmul(h2_new, w3) + b3
h3 = torch.nn.ReLU()(l3)

w4 = torch.randn(size=(128, 10), requires_grad=True)
b4 = torch.randn(size=(1, 10), requires_grad=True)
l4 = torch.matmul(h3, w4) + b4
h4 = torch.nn.ReLU()(l4)

loss = torch.nn.functional.binary_cross_entropy_with_logits(y, h4)
loss.backward()

print(w2.grad) #we can see that gradient computing is cut off at the h2_new, by treating h2_new as a leaf node.
```
