```python
import os
import numbers
```

```python
class Value:
    
    def __init__(self, value, label="", children=()):
        assert(isinstance(value, numbers.Number) and not isinstance(value, bool))
        self.value = value
        self.label = label
        self.children = children
        self.grad = 0
        self._backward = lambda: None
        
    def __add__(self, other):
        assert(isinstance(other, Value) or (isinstance(value, numbers.Number) and not isinstance(value, bool)))
        out = Value(self.value + other.value, children = (self, other))
        def _backward():
            self.grad += out.grad
            other.grad += out.grad
        out._backward = _backward
        
        return out
        
    def __mul__(self, other):
        assert(isinstance(other, Value) or (isinstance(value, numbers.Number) and not isinstance(value, bool)))
        out = Value(self.value * other.value, children = (self, other))
        
        def _backward():
            self.grad += other.value * out.grad
            other.grad += self.value * out.grad
        out._backward = _backward
        
        return out
    
    def backward(self):
        self.grad = 1
        
        topo = []
        visited = set()
        def build_topo(node):
            if node not in visited:
                visited.add(node)
                for child in node.children:
                    build_topo(child)
                topo.append(node)
        
        build_topo(self)
        print(topo)
        
        for node in reversed(topo):
            node._backward()
        
    def __str__(self):
        return f"Value: {self.value} & Label: {self.label}"
    
    def __repr__(self):
        return self.__str__()
    
```

```python
a = Value(10, "a")
b = Value(20, "b")
c = a+b
d = Value(30, "d")
e = c*d
e.backward()
print(e.grad)
print(d.grad)
print(c.grad)
print(b.grad)
print(a.grad)
```
